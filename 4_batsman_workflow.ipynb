```python
from langgraph.graph import StateGraph , START, END
from typing import TypedDict
```

```python
class BatsmanState(TypedDict):
  runs:int
  balls: int
  fours:int
  sixes:int

  sr:float
  bpb:float
  boundary_percent:float
  summary:str
```

```python
def calculate_src(state:BatsmanState) -> dict:
  sr = (state["runs"] / state["balls"]) * 100 if state["balls"] else 0


  return {'sr':sr}
```

```python
def calculate_bpb(state:BatsmanState)->dict:
  bpb = state['balls']/(state['fours'] + state['sixes'])


  return {'bpb':bpb}
```

```python
def calculate_boundary_percent(state:BatsmanState) -> dict:

  boundary_percent = (((state['fours']*4 + state['sixes']*6))/state['runs'])*100

  

  return {'boundary_percent':boundary_percent}


```

```python
def summary(state:BatsmanState) -> dict:
  summary = f"""
Strike rate = {state['sr']} \n
Balls per boundary = {state['bpb']} \n
Boundary percent = {state['boundary_percent']} \n

"""
  

  return {'summary':summary}
```

```python
graph = StateGraph(BatsmanState)

# ADD Nodes
graph.add_node('calculate_src',calculate_src)
graph.add_node('calculate_bpb',calculate_bpb)
graph.add_node('calculate_boundary_percent',calculate_boundary_percent)
graph.add_node('summary',summary)

# ADD Edges
graph.add_edge(START, 'calculate_src')
graph.add_edge(START, 'calculate_bpb')
graph.add_edge(START,'calculate_boundary_percent')

graph.add_edge('calculate_src','summary')
graph.add_edge('calculate_bpb','summary')
graph.add_edge('calculate_boundary_percent','summary')
graph.add_edge('summary',END)

workflow = graph.compile()
 
```

```python
initial_state = {
  'runs':100,
  'balls':50,
  'fours':8,
  'sixes':6

}

final_response = workflow.invoke(initial_state)

print(final_response)
```
